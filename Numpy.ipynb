```python
import numpy as np
import pandas as pd


# 1. Load Dataset
df = pd.read_csv("WineQT.csv")

print("DATASET")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Column Names:", df.columns.tolist())


# 2. Filter Rows
filtered = df[
    (df["citric acid"] > 0.5) |
    (df["volatile acidity"] < 0.3)
]

print("\nFILTERED ROWS")
print("Number of filtered rows:", filtered.shape[0])

percentage = (filtered.shape[0] / df.shape[0]) * 100

print("Percentage:", round(percentage, 2), "%")


# 3. Z-Score Normalization
alcohol = df["alcohol"].to_numpy()
ph = df["pH"].to_numpy()

alcohol_z = (alcohol - np.mean(alcohol)) / np.std(alcohol)
ph_z = (ph - np.mean(ph)) / np.std(ph)

z_scores = np.column_stack((alcohol_z, ph_z))

print("\nZ-SCORE")
print(z_scores[:5])


# 4. Group By Quality
print("\nGROUP BY QUALITY")

for quality, data in df.groupby("quality"):

    mean_acidity = data["volatile acidity"].mean()
    median_alcohol = data["alcohol"].median()

    print(
        "Quality:", quality,
        "Mean Volatile Acidity:", round(mean_acidity, 4),
        "Median Alcohol:", round(median_alcohol, 2)
    )


# 5. Correlation
correlation = np.corrcoef(
    df["alcohol"],
    df["quality"]
)[0, 1]

print("\nCORRELATION")
print("Alcohol and Quality:",
      round(correlation, 4))


# 6. Acidity Ratio
df["Acidity Ratio"] = (
    df["fixed acidity"] /
    df["volatile acidity"]
)

print("\nACIDITY RATIO")
print(df["Acidity Ratio"].head())


# 7. Convert to NumPy Array
final_array = df.to_numpy()

print("\nFINAL ARRAY")
print("Shape:", final_array.shape)
print(final_array[:5])
```
